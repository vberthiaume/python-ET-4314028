```python
class NonIntArgumentException(Exception):
    pass

# Python code​​​​​‌​​​​​​​‌‌​‌‌‌‌​‌‌‌​‌​​​‌‌ below
def handleNonIntArguments(func):
    def wrapper(*args):
        for arg in args:
            if not isinstance(arg, int):
                raise NonIntArgumentException

        return func (*args)
    return wrapper
```
