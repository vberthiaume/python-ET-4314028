```python
import json

json_string = '{"a": "apple", "b": "bear", "c": "cat"}'

try:
    data = json.loads(json_string)
except json.JSONDecodeError:
    print('Could not parse JSON!')
else:
	# data is available here because python only makes scopes at the function, class, or module level
    print(f'data of type {type(data)} contains {data}')
```
