```python
class Dog:
	# properties defined at the class level are class attributes, like static in c++
    _legs = 4 # convention is to have _ before class attributes

    def __init__(self, name):
	    # properties defined in init are instance attributes, you can only access them through an instance, like regular member variables
        self.name = name
        
    def speak(self):
        print(self.name + ' says: Bark!')

myDog = Dog('Rover')
# this will actually add an instance attribute to the class
myDog._legs = 3
print(myDog._legs)

# without changing the class attribule lol
print(Dog._legs)
```
