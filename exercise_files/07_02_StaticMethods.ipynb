## Static and Instance Methods

```python
class WordSet:
    def __init__(self):
        self.words = set()

    # static method doesn't take self as parameter
    def cleanText(text):
        text = text.replace('!', '').replace('.', '').replace(',', '').replace('\'', '')
        return text.lower()

    # instance methods do take self as parameter
    def addText(self, text):
        # static method needs the class name to be called
        text = WordSet.cleanText(text)
        for word in text.split():
            self.words.add(word)

    
    
        
wordSet = WordSet()

wordSet.addText('Hi, I\'m Ryan! Here is a sentence I want to add!')
wordSet.addText('Here is another sentence I want to add.')

print(wordSet.words)
```

```python
class WordSet:
    # static class attribute
    replacePuncs = ['!', '.', ',', '\'']
    def __init__(self):
        self.words = set()

    # optional static method decorator to be able to call the method on an instance
    @staticmethod
    def cleanText(text):    # static method doesn't take self as parameter
        for punc in WordSet.replacePuncs:
            text = text.replace(punc, '')   # chaining functions
        return text.lower()

# instance methods do take self as parameter
    def addText(self, text):
        # static method needs the class name to be called
        # text = WordSet.cleanText(text)
        # unless you have the @staticmethod above, then you cal call it right on the instance
        text = self.cleanText(text)
        for word in text.split():
            self.words.add(word)

wordSet = WordSet()

wordSet.addText('Hi, I\'m Ryan! Here is a sentence I want to add!')
wordSet.addText('Here is another sentence I want to add.')

print(wordSet.words)
```

### Decorators 

```python
class WordSet:
    replacePuncs = ['!', '.', ',', '\'']
    def __init__(self):
        self.words = set()
        
    def addText(self, text):
        # this is only possible because cleanText is declared as staticmethod below
        text = self.cleanText(text)
        for word in text.split():
            self.words.add(word)

    # optional decorator to be able to do the above trick
    @staticmethod
    def cleanText(text):
        # chaining functions
        for punc in WordSet.replacePuncs:
            text = text.replace(punc, '')
        return text.lower()
    
        
wordSet = WordSet()

wordSet.addText('Hi, I\'m Ryan! Here is a sentence I want to add!')
wordSet.addText('Here is another sentence I want to add.')

print(wordSet.words)
```
