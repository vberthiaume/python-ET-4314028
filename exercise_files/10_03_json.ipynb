```python
import json
from json import JSONDecodeError, JSONEncoder
```

## JSON

### Loading JSON

```python
# make a json string
jsonString = '{"a": "apple", "b": "bear", "c": "cat"}'

# and convert it to a json dict. It's common for json files to be wonky so wrap in try block
try:
    dict = json.loads(jsonString)
except JSONDecodeError:
    print('Could not parse JSON!')

print (f'dict of type {type(dict)} contains {dict}')
```

### Dumping JSON

```python
# easy to just dump a native python dic
pythonDict = {'a': 'apple', 'b': 'bear', 'c': 'cat',}
json.dumps(pythonDict)
```

### Custom JSON Decoders

```python
# but if you have a class
class Animal:
    def __init__(self, name):
        self.name = name

# you need a custom encoder that extends the default JSONEncoder
class AnimalEncoder(JSONEncoder):
    # you only need to override this method
    def default(self, o): # o is the object that needs to be encoded in json
        if type(o) == Animal:
            return o.name
        return super().default(o)
    
pythonDict = {'a': Animal('aardvark'), 'b': Animal('bear'), 'c': Animal('cat'),}
json.dumps(pythonDict, cls=AnimalEncoder) # and cls is where you give the custom encoder
```
