## CSV

### Reading

```python
with open('10_02_us.csv', 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    for row in reader:
        print(row)
```

```python
with open('10_02_us.csv', 'r') as f:
    reader = csv.reader(f, delimiter='\t') # delimiter is comma by default
    next(reader) # this will skip a line, it keeps a bookmark so you can call multiple times
    for row in reader[1:]:  # you can iterate over reador or convert to a list like here
        print(row)
```

```python
with open('10_02_us.csv', 'r') as f:
    # and this is a built-in way of skipping the header; it'll just parse it properly in a dict
    reader = csv.DictReader(f, delimiter='\t')
    for row in reader:
        print(row)
```

### Filtering data

```python
with open('10_02_us.csv', 'r') as f:
    data = list(csv.DictReader(f, delimiter='\t'))
```

```python
primes = []
for number in range(2, 99999):
    for factor in range(2, int(number**0.5) + 1):
        if number % factor == 0:
            break
    else:
        primes.append(number)
```

```python
data = [row for row in data if int(row['postal code']) in primes and row['state code'] == 'MA']
len(data)
```

### Writing

```python
with open('10_02_ma_prime.csv', 'w') as f:
    writer = csv.writer(f)
    for row in data:
        writer.writerow([row['place name'], row['county']])
        
        
```
